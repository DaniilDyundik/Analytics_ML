```python

def columnValues(series):
    total_rows = len(series)
    
    unique_count = series.nunique()
    zeros_count = (series == 0).sum() 
    nan_count = series.isna().sum()    
    
    # Рассчитываем проценты
    unique_percent = round((unique_count / total_rows * 100), 1)
    zeros_percent = round((zeros_count / total_rows * 100), 1)
    nan_percent = round((nan_count / total_rows * 100), 1)
    
    # Преобразуем dtype в строку
    data_type = str(series.dtype)
    
    # Формируем строку
    row = [unique_count, f"{unique_percent}%",
           zeros_count, f"{zeros_percent}%",
           nan_count, f"{nan_percent}%",
           data_type]  # теперь это строка
    return row

data = []
for column in df.columns:
    data.append([column] + columnValues(df[column]))

print(tabulate(data, headers=['Column', 'Unique', '% of_Unique', 'Zeros', '% of_Zeros', 'NaNs', '% of_NaNs', 'dtype'],
               tablefmt='orgtbl'))
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from statsmodels.tsa.statespace.sarimax import SARIMAX
from sklearn.metrics import mean_absolute_error, r2_score, mean_absolute_percentage_error
import warnings
warnings.filterwarnings('ignore')

# Маска для выкупа 
mask = (df_clean['Отменено'] == 'Нет') & \
       (df_clean['Статус'] == 'Доставлен') & \
       (df_clean['НовыйСтатус'] == 'Доставлен')


# Данные по выкупленным товарам 
detstvo = df_clean[mask & (df_clean['Группа2'] == 'ИГРУШКИ')].groupby('Date')['КоличествоПроданоКлиенту'].sum()

train = detstvo[:-7]
test = detstvo[-7:]

# Модель
model = SARIMAX(train, order=(1,0,0), seasonal_order=(2,1,0,7))
results = model.fit(disp=False)
forecast = results.forecast(steps=7)

# График
plt.figure(figsize=(12,5))

# Полный ряд 
plt.plot(detstvo.index, detstvo.values, 'b-', linewidth=1.5, label='Фактические данные')

# Прогноз на последние 7 дней
plt.plot(forecast.index, forecast.values, 'r-', linewidth=2, label='Прогноз SARIMA')

plt.title('Детские товары: прогноз на 7 дней')
plt.xlabel('Дата')
plt.ylabel('Продажи')
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()

# Метрики
mae = mean_absolute_error(test, forecast)
r2 = r2_score(test, forecast)
mape = mean_absolute_percentage_error(test, forecast) * 100  # умножаем на 100 для процентов

print(f'MAE: {mae:.2f}')
print(f'MAPE: {mape:.2f}%')
print(f'R²: {r2:.2f}')
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from statsmodels.tsa.statespace.sarimax import SARIMAX
from sklearn.metrics import mean_absolute_error, mean_absolute_percentage_error, r2_score
from scipy import stats
import warnings
warnings.filterwarnings('ignore')

# ПОДГОТОВКА ДАННЫХ
mask = (df_clean['Отменено'] == 'Нет') & \
       (df_clean['Статус'] == 'Доставлен') & \
       (df_clean['НовыйСтатус'] == 'Доставлен')

daily_data = df_clean[mask & (df_clean['Группа2'] == 'ИГРУШКИ')].groupby('Date').agg(
    sales=('КоличествоПроданоКлиенту', 'sum'),
    sale_percent=('sale_percent', 'mean')
)

daily_data.index = pd.to_datetime(daily_data.index)
daily_data = daily_data.sort_index()
daily_data['sale_percent_lag1'] = daily_data['sale_percent'].shift(1)
daily_data = daily_data.dropna()

# КРОСС-ВАЛИДАЦИЯ 
test_size, min_train = 7, 60
n_splits = len(daily_data) - min_train - test_size + 1

results_with, results_without = [], []

for i in range(n_splits):
    train_end = min_train + i
    test_start, test_end = train_end, train_end + test_size
    
    train = daily_data.iloc[:train_end]
    test = daily_data.iloc[test_start:test_end]
    
    # Прогноз sale_percent
    model_sp = SARIMAX(train['sale_percent'], order=(1,0,0), seasonal_order=(1,1,0,7))
    res_sp = model_sp.fit(disp=False)
    fc_sp = res_sp.forecast(steps=test_size)
    
    exog_test = pd.DataFrame(index=test.index)
    exog_test['sale_percent_lag1'] = fc_sp.shift(1)
    exog_test['sale_percent_lag1'].iloc[:1] = train['sale_percent'].iloc[-1:].values
    
    # Модель с sale_percent
    model_s = SARIMAX(train['sales'], exog=train[['sale_percent_lag1']], order=(1,0,0), seasonal_order=(2,1,0,7))
    res_s = model_s.fit(disp=False)
    fc_s = res_s.forecast(steps=test_size, exog=exog_test)
    
    results_with.append({
        'train_size': len(train),
        'mape': mean_absolute_percentage_error(test['sales'], fc_s) * 100,
        'mae': mean_absolute_error(test['sales'], fc_s),
        'r2': r2_score(test['sales'], fc_s)
    })
    
    # Модель без учета фактора
    model_0 = SARIMAX(train['sales'], order=(1,0,0), seasonal_order=(2,1,0,7))
    res_0 = model_0.fit(disp=False)
    fc_0 = res_0.forecast(steps=test_size)
    
    results_without.append({
        'train_size': len(train),
        'mape': mean_absolute_percentage_error(test['sales'], fc_0) * 100,
        'mae': mean_absolute_error(test['sales'], fc_0),
        'r2': r2_score(test['sales'], fc_0)
    })

#  РЕЗУЛЬТАТЫ
df_with = pd.DataFrame(results_with)
df_without = pd.DataFrame(results_without)


print(f"РЕЗУЛЬТАТЫ (n={len(df_with)} окон)")

print(f"С sale_percent (lag 1):  MAPE = {df_with['mape'].mean():.1f}% ± {df_with['mape'].std():.1f}%")
print(f"Без экзогена:            MAPE = {df_without['mape'].mean():.1f}% ± {df_without['mape'].std():.1f}%")
print(f"Разница:                 {df_without['mape'].mean() - df_with['mape'].mean():+.1f}%")

t_stat, p_value = stats.ttest_rel(df_with['mape'], df_without['mape'])
print(f"\nПарный t-тест: p-value = {p_value:.4f} {' значимо' if p_value < 0.05 else ' не значимо'}")

# ГРАФИК
plt.figure(figsize=(10, 6))
plt.plot(df_with['train_size'], df_with['mape'], 'b-o', label='С учетом скидки')
plt.plot(df_without['train_size'], df_without['mape'], 'r-s', label='Без учета скидки')
plt.axhline(y=df_with['mape'].mean(), color='b', linestyle='--', alpha=0.5)
plt.axhline(y=df_without['mape'].mean(), color='r', linestyle='--', alpha=0.5)
plt.xlabel('Размер train (дней)')
plt.ylabel('MAPE, %')
plt.title(f'Зависимость MAPE от размера обучающей выборки')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
from catboost import CatBoostRegressor, Pool
from sklearn.metrics import mean_absolute_percentage_error, mean_absolute_error, r2_score
import matplotlib.pyplot as plt

# ПОДГОТОВКА ДАННЫХ
mask = (df_clean['Отменено'] == 'Нет') & \
       (df_clean['Статус'] == 'Доставлен') & \
       (df_clean['НовыйСтатус'] == 'Доставлен')

daily_data = df_clean[mask & (df_clean['Группа2'] == 'ИГРУШКИ')].groupby('Date').agg(
    sales=('КоличествоПроданоКлиенту', 'sum'),
    discount_share=('real_discount', 'mean'),
    margin=('Маржа', 'mean')
).reset_index()

daily_data['date'] = pd.to_datetime(daily_data['Date'])
daily_data = daily_data.sort_values('date')

# ФУНКЦИЯ СОЗДАНИЯ ПРИЗНАКОВ (ИЗНАЧАЛЬНАЯ)
def create_features_honest(data, train_end_idx):
    data = data.copy()
    
    # Лаги продаж
    for lag in [1, 2, 3, 7, 14]:
        data[f'sales_lag{lag}'] = data['sales'].shift(lag)
    
    # Скидка
    data['discount_lag1'] = data['discount_share'].shift(1)
    data['discount_lag5'] = data['discount_share'].shift(5)
    #
    # Маржа
    data['margin_lag1'] = data['margin'].shift(1)
    
    # Скользящие средние
    data['sales_ma7'] = data['sales'].shift(1).rolling(7, min_periods=1).mean()
    
    # Календарные
    data['dayofweek'] = data['date'].dt.dayofweek
    data['is_weekend'] = data['dayofweek'].isin([5, 6]).astype(int)
    data['month'] = data['date'].dt.month
    
    return data

# РАЗДЕЛЕНИЕ ДО СОЗДАНИЯ ПРИЗНАКОВ 
train_size = len(daily_data) - 7
train_raw = daily_data.iloc[:train_size].copy()
test_raw = daily_data.iloc[train_size:].copy()

train = create_features_honest(train_raw, train_size)
train = train.dropna()

print(f"Train: {len(train)} дней")
print(f"Test: {len(test_raw)} дней")

# ПРИЗНАКИ 
features = [
    'sales_lag1', 'sales_lag2', 'sales_lag3', 'sales_lag7', 'sales_lag14',
    'discount_lag1', 'discount_lag5',
    'margin_lag1',
    'sales_ma7',
    'dayofweek', 'is_weekend', 'month'
]

cat_features = ['dayofweek', 'month', 'is_weekend']

X_train = train[features]
y_train = train['sales']

model = CatBoostRegressor(
    iterations=300,
    learning_rate=0.03,
    depth=5,
    loss_function='RMSE',
    eval_metric='MAPE',
    random_seed=42,
    early_stopping_rounds=20,
    verbose=50
)

train_pool = Pool(X_train, y_train, cat_features=cat_features)
model.fit(train_pool)

# ========== 6. ПРОГНОЗ НА TEST (динамический) ==========
test_predictions = []
history_sales = list(train['sales'].values)

for i in range(len(test_raw)):
    current_features = {}
    
    # Лаги продаж
    current_features['sales_lag1'] = history_sales[-1]
    current_features['sales_lag2'] = history_sales[-2] if len(history_sales) >= 2 else history_sales[-1]
    current_features['sales_lag3'] = history_sales[-3] if len(history_sales) >= 3 else history_sales[-1]
    current_features['sales_lag7'] = history_sales[-7] if len(history_sales) >= 7 else history_sales[-1]
    current_features['sales_lag14'] = history_sales[-14] if len(history_sales) >= 14 else history_sales[-1]
    
    # MA7
    current_features['sales_ma7'] = np.mean(history_sales[-7:]) if len(history_sales) >= 7 else np.mean(history_sales)
    
    # Скидка
    current_features['discount_lag1'] = test_raw.iloc[i-1]['discount_share'] if i > 0 else train_raw.iloc[-1]['discount_share']
    current_features['discount_lag5'] = test_raw.iloc[i-5]['discount_share'] if i >= 5 else train_raw.iloc[-5+i]['discount_share'] if -5+i < 0 else 0
    
    # Маржа
    current_features['margin_lag1'] = test_raw.iloc[i-1]['margin'] if i > 0 else train_raw.iloc[-1]['margin']
    
    # Календарные
    current_date = test_raw.iloc[i]['date']
    current_features['dayofweek'] = current_date.dayofweek
    current_features['is_weekend'] = 1 if current_date.dayofweek in [5, 6] else 0
    current_features['month'] = current_date.month
    
    # Прогноз
    X_pred = pd.DataFrame([current_features])[features]
    pred = model.predict(X_pred)[0]
    test_predictions.append(pred)
    
    history_sales.append(pred)

# ========== 7. МЕТРИКИ ==========
y_test = test_raw['sales'].values
pred = np.array(test_predictions)

mae = mean_absolute_error(y_test, pred)
mape = mean_absolute_percentage_error(y_test, pred) * 100
r2 = r2_score(y_test, pred)

print(f'\n=== РЕЗУЛЬТАТЫ CATBOOST (ИЗНАЧАЛЬНАЯ КОНФИГУРАЦИЯ) ===')
print(f'MAE: {mae:.2f}')
print(f'MAPE: {mape:.2f}%')
print(f'R²: {r2:.2f}')

print("\nСравнение фактических и прогнозных продаж:")
comparison = pd.DataFrame({
    'Дата': test_raw['date'].dt.strftime('%Y-%m-%d').values,
    'Факт': y_test.round(0),
    'Прогноз': pred.round(0),
    'Отклонение': (pred - y_test).round(0)
})
print(comparison.to_string(index=False))

# ========== 8. ГРАФИК ==========
fig, axes = plt.subplots(2, 1, figsize=(14, 10))

axes[0].plot(daily_data['date'], daily_data['sales'], 'b-', linewidth=1.5, label='Фактические продажи')
axes[0].plot(test_raw['date'], pred, 'red', linewidth=2, label='Прогноз CatBoost', marker='o', markersize=6)
axes[0].axvline(x=test_raw['date'].iloc[0], color='gray', linestyle='--', alpha=0.7, label='Начало прогноза')
axes[0].set_title(f'ИГРУШКИ: CatBoost ИЗНАЧАЛЬНЫЙ (MAPE={mape:.1f}%, R²={r2:.2f})', fontsize=14)
axes[0].set_xlabel('Дата')
axes[0].set_ylabel('Продажи (кол-во)')
axes[0].grid(True, alpha=0.3)
axes[0].legend()

axes[1].plot(test_raw['date'], y_test, 'g-o', linewidth=2, label='Факт', markersize=8)
axes[1].plot(test_raw['date'], pred, 'red', linewidth=2, label='Прогноз', marker='s', markersize=8)
axes[1].set_title('Тестовый период (7 дней) — детально', fontsize=14)
axes[1].set_xlabel('Дата')
axes[1].set_ylabel('Продажи (кол-во)')
axes[1].grid(True, alpha=0.3)
axes[1].legend()

for i, (date, fact, pr) in enumerate(zip(test_raw['date'], y_test, pred)):
    axes[1].annotate(f'{fact:.0f}', (date, fact), textcoords="offset points", xytext=(0, 10), ha='center', fontsize=9, color='green')
    axes[1].annotate(f'{pr:.0f}', (date, pr), textcoords="offset points", xytext=(0, -15), ha='center', fontsize=9, color='red')

plt.tight_layout()
plt.show()

# ========== 9. ВАЖНОСТЬ ПРИЗНАКОВ ==========
print(f'\n=== ВАЖНОСТЬ ПРИЗНАКОВ (изначальная) ===')
feature_importance = model.get_feature_importance(prettified=True)
print(feature_importance.head(12))

plt.figure(figsize=(10, 6))
top_features = feature_importance.head(10)
plt.barh(top_features['Feature Id'], top_features['Importances'])
plt.xlabel('Важность')
plt.title('Top-10 важных признаков CatBoost (изначальная конфигурация)')
plt.gca().invert_yaxis()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
